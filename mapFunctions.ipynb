```python
import pandas as pd
```

```python
df=pd.read_csv('StudentDataset.csv')
```

```python
df.describe()
```

```python
def pass_fail(score):
  if score >80:
    return 'Pass'
  else:
    return 'Fail'
df['Math_Result']=df['Math_Score'].apply(pass_fail)
df[['Student_Name','Math_Score','Math_Result']].head(10)
```

```python
df['Science_Result']=df['Science_Score'].apply(lambda x: 'pass' if x>80 else 'fail')
df[['Science_Score','Science_Result']].head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(30, 5))

plt.plot(df["Student_Name"], df["Math_Score"], marker="o", label="Math")
plt.plot(df["Student_Name"], df["Science_Score"], marker="o", label="Science")
plt.plot(df["Student_Name"], df["English_Score"], marker="o", label="English")

plt.title("Student Performance by Subject")
plt.xlabel("Student")
plt.ylabel("Score")

plt.legend()
plt.show()
```
